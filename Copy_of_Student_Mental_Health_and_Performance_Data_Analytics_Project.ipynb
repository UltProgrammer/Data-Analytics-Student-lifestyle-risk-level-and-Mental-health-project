## 1. Data Loading and Initial Inspection

First, we'll load the dataset into a pandas DataFrame and perform a preliminary inspection to understand its structure, data types, and check for any missing values.

```python
import pandas as pd

# Load the dataset
file_path = '/content/student_mental_health_burnout_1M.csv'
df = pd.read_csv(file_path)

# Display the first 5 rows of the DataFrame
display(df.head())
```

```python
# Display basic information about the DataFrame, including data types and non-null values
display(df.info())
```

```python
# Display descriptive statistics for numerical columns
display(df.describe())

# Display descriptive statistics for all columns, including categorical ones
display(df.describe(include='all'))
```

```python
# Check for missing values in each column
display(df.isnull().sum().sort_values(ascending=False))
```

## 2. Data Cleaning and Preprocessing

Based on the initial inspection, we've identified missing values. We will now address these and prepare the data for further analysis.

```python
# Drop rows with any missing values
df_cleaned = df.dropna().copy()

# Verify that there are no more missing values
print("Missing values after dropping rows:")
display(df_cleaned.isnull().sum().sort_values(ascending=False))

print(f"Original DataFrame shape: {df.shape}")
print(f"Cleaned DataFrame shape: {df_cleaned.shape}")
```

### 2.1 Categorical Feature Encoding

We will encode categorical variables to numerical representations suitable for machine learning models. 'gender' will be one-hot encoded, and 'risk_level' will be label encoded as it has a clear ordinal relationship (Low < Medium < High).

```python
# One-hot encode 'gender'
df_encoded = pd.get_dummies(df_cleaned, columns=['gender'], drop_first=True)

# Label encode 'risk_level'
risk_level_mapping = {'Low': 0, 'Medium': 1, 'High': 2}
df_encoded['risk_level_encoded'] = df_encoded['risk_level'].map(risk_level_mapping)

# Drop the original 'risk_level' column as it's now encoded
df_encoded = df_encoded.drop('risk_level', axis=1)

print("DataFrame after encoding categorical features:")
display(df_encoded.head())
```

### 2.2 Numerical Feature Transformation to Integers

We will now transform several numerical features to integers based on specific rounding rules:
*   `academic_year` and `study_hours_per_day`: Round up to the nearest integer.
*   `sleep_hours`: Round down to the nearest integer.
*   Other float columns: Round to the nearest integer.

```python
import numpy as np

df_processed = df_encoded.copy()

# Round up academic_year and study_hours_per_day
df_processed['academic_year'] = np.ceil(df_processed['academic_year']).astype(int)
df_processed['study_hours_per_day'] = np.ceil(df_processed['study_hours_per_day']).astype(int)

# Round down sleep_hours
df_processed['sleep_hours'] = np.floor(df_processed['sleep_hours']).astype(int)

# Round other float columns to the nearest integer
columns_to_round_nearest = [
    'exam_pressure', 'academic_performance', 'stress_level', 'anxiety_score',
    'depression_score', 'physical_activity', 'social_support', 'screen_time',
    'internet_usage', 'financial_stress', 'family_expectation', 'burnout_score',
    'mental_health_index', 'dropout_risk'
]

for col in columns_to_round_nearest:
    if col in df_processed.columns:
        df_processed[col] = np.round(df_processed[col]).astype(int)

print("DataFrame after converting specified features to integers:")
display(df_processed.head())
display(df_processed.info())
```

### 3.2.1 Strong Correlation Analysis

Let's visualize only the feature pairs with an absolute correlation greater than 0.5 to highlight the most significant relationships.

```python
# Create a mask for correlations above 0.5 (absolute value)
strong_correlations = correlation_matrix[abs(correlation_matrix) > 0.5].fillna(0)

# Set up the matplotlib figure
plt.figure(figsize=(18, 15))
sns.heatmap(strong_correlations, annot=True, cmap='coolwarm', fmt='.2f', linewidths=.5)
plt.title('Strong Correlation Matrix (Absolute Correlation > 0.5)')
plt.show()
```

### 3.2.2 Visualizing Strongly Correlated Feature Pairs

To further understand the strong relationships identified in the correlation matrix, we will create scatter plots for each pair of features where the absolute correlation coefficient is greater than 0.5. This helps visualize the nature and direction of these relationships.

```python
import math

columns = strong_correlations.columns

# Count how many strong correlations there are to determine subplot grid size
num_strong_correlations = 0
for i in range(len(columns)):
    for j in range(i + 1, len(columns)): # Iterate through upper triangle
        correlation_value = strong_correlations.loc[columns[i], columns[j]]
        if abs(correlation_value) > 0.5:
            num_strong_correlations += 1

if num_strong_correlations == 0:
    print("No feature pairs found with absolute correlation greater than 0.5 for individual scatter plots.")
else:
    # Determine grid size for subplots
    n_cols = 4 # Number of columns in the subplot grid
    n_rows = math.ceil(num_strong_correlations / n_cols)

    plt.figure(figsize=(n_cols * 5, n_rows * 4)) # Adjust figure size dynamically
    plot_idx = 0

    for i in range(len(columns)):
        for j in range(i + 1, len(columns)): # Iterate through upper triangle
            col1 = columns[i]
            col2 = columns[j]
            correlation_value = strong_correlations.loc[col1, col2]

            if abs(correlation_value) > 0.5:
                plot_idx += 1
                plt.subplot(n_rows, n_cols, plot_idx)
                sns.scatterplot(x=df_processed[col1], y=df_processed[col2], alpha=0.6)
                plt.title(f'{col1} vs {col2} (Corr: {correlation_value:.2f})', fontsize=10)
                plt.xlabel(col1, fontsize=8)
                plt.ylabel(col2, fontsize=8)
                plt.grid(True, linestyle='--', alpha=0.7)

    plt.tight_layout()
    plt.show()
    print(f"Generated {num_strong_correlations} scatter plots for strongly correlated feature pairs in a single figure.")
```

## 4. Machine Learning Model Building

We will now proceed to build machine learning models to predict important outcomes. Given the nature of the data, predicting the `risk_level_encoded` (Low, Medium, High) is a relevant classification task. We will also explore predicting `dropout_risk` as a regression task.

### 4.1 Data Preparation for Modeling

First, we need to define our features (X) and target variable (y). Then, we will split the dataset into training and testing sets to evaluate our models effectively.

```python
from sklearn.model_selection import train_test_split

# Define features (X) and target (y) for classification
# We'll drop the original gender columns, risk_level_encoded (as it's the target),
# and dropout_risk (as it's another target we're predicting separately) from X.
X = df_processed.drop(columns=['risk_level_encoded', 'dropout_risk', 'gender_Male', 'gender_Other'])
y = df_processed['risk_level_encoded']

# Split the data into training and testing sets (70% train, 30% test)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)

print(f"Shape of X_train: {X_train.shape}")
print(f"Shape of X_test: {X_test.shape}")
print(f"Shape of y_train: {y_train.shape}")
print(f"Shape of y_test: {y_test.shape}")
```

### 4.2 Model Training - Classification (Predicting Risk Level)

We will train a couple of common classification models: Logistic Regression and Random Forest Classifier. These models are robust and provide good baselines for comparison.

```python
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report
from sklearn.model_selection import cross_val_score, StratifiedKFold
from sklearn.preprocessing import StandardScaler # New: Import StandardScaler

# Initialize and train Logistic Regression model
# Increased max_iter for convergence with large dataset

# New: Scale the data for Logistic Regression
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
X_scaled = scaler.transform(X) # Also scale the full X for cross-validation

# Changed solver to 'saga' for better convergence with large datasets, supports n_jobs=-1
log_reg_model = LogisticRegression(max_iter=10000, random_state=42, n_jobs=-1, solver='saga')
log_reg_model.fit(X_train_scaled, y_train) # Use scaled data here

# Make predictions
y_pred_log_reg = log_reg_model.predict(X_test_scaled) # Use scaled data here

# Evaluate the model
print("Logistic Regression Model Evaluation:")
print(f"Accuracy: {accuracy_score(y_test, y_pred_log_reg):.4f}")
print(classification_report(y_test, y_pred_log_reg))

# Perform cross-validation
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
# New: Pass scaled X for cross_val_score as well
log_reg_cv_scores = cross_val_score(log_reg_model, X_scaled, y, cv=cv, scoring='accuracy', n_jobs=-1)
print(f"\nLogistic Regression Cross-Validation Accuracy: {log_reg_cv_scores.mean():.4f} (+/- {log_reg_cv_scores.std() * 2:.4f})")
```

```python
# Initialize and train Random Forest Classifier model
# Increased n_estimators and added n_jobs for faster training and better performance
rf_model = RandomForestClassifier(n_estimators=100, random_state=42, n_jobs=-1)
rf_model.fit(X_train, y_train)

# Make predictions
y_pred_rf = rf_model.predict(X_test)

# Evaluate the model
print("\nRandom Forest Classifier Model Evaluation:")
print(f"Accuracy: {accuracy_score(y_test, y_pred_rf):.4f}")
print(classification_report(y_test, y_pred_rf))

# Perform cross-validation
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
rf_cv_scores = cross_val_score(rf_model, X, y, cv=cv, scoring='accuracy', n_jobs=-1)
print(f"\nRandom Forest Cross-Validation Accuracy: {rf_cv_scores.mean():.4f} (+/- {rf_cv_scores.std() * 2:.4f})")
```

### 4.3 Model Training - Regression (Predicting Dropout Risk)

For the second model, we'll predict 'dropout_risk', which is a regression problem. We'll use a RandomForest Regressor.

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.model_selection import train_test_split

# Define features (X_reg) and target (y_reg) for regression
X_reg = df_processed.drop(columns=['dropout_risk', 'risk_level_encoded', 'gender_Male', 'gender_Other'])
y_reg = df_processed['dropout_risk']

# Split the data into training and testing sets for regression (70% train, 30% test)
X_train_reg, X_test_reg, y_train_reg, y_test_reg = train_test_split(X_reg, y_reg, test_size=0.3, random_state=42)

print(f"Shape of X_train_reg: {X_train_reg.shape}")
print(f"Shape of X_test_reg: {X_test_reg.shape}")

# Initialize and train Random Forest Regressor model
rf_reg_model = RandomForestRegressor(n_estimators=100, random_state=42, n_jobs=-1)
rf_reg_model.fit(X_train_reg, y_train_reg)

# Make predictions
y_pred_rf_reg = rf_reg_model.predict(X_test_reg)

# Evaluate the model
print("\nRandom Forest Regressor Model Evaluation (Predicting Dropout Risk):")
print(f"Mean Absolute Error: {mean_absolute_error(y_test_reg, y_pred_rf_reg):.4f}")
print(f"Mean Squared Error: {mean_squared_error(y_test_reg, y_pred_rf_reg):.4f}")
print(f"R-squared: {r2_score(y_test_reg, y_pred_rf_reg):.4f}")
```

## 5. Interactive Prediction Interface

We will now create three distinct interactive interfaces to input feature values and obtain predictions from our trained models. Each panel will focus on a specific set of predictions.

```python
import ipywidgets as widgets
from IPython.display import display, clear_output
import pandas as pd

# Get feature names for classification model (X_train is defined earlier)
classification_features = X_train.columns.tolist()

# Define features for Dropout Risk regression (replicate X_reg definition)
# These features exclude dropout_risk, risk_level_encoded, gender_Male, gender_Other
regression_features = df_processed.drop(columns=['dropout_risk', 'risk_level_encoded', 'gender_Male', 'gender_Other']).columns.tolist()

# Define features for Mental Health Index regression (replicate X_mhi definition)
# These features exclude mental_health_index, burnout_score, dropout_risk, risk_level_encoded, gender_Male, gender_Other
mhi_input_features = df_processed.drop(columns=['mental_health_index', 'burnout_score', 'dropout_risk', 'risk_level_encoded', 'gender_Male', 'gender_Other']).columns.tolist()

# Define features for Burnout Score regression (replicate X_burnout definition)
# These features exclude burnout_score, mental_health_index, dropout_risk, risk_level_encoded, gender_Male, gender_Other
burnout_input_features = df_processed.drop(columns=['burnout_score', 'mental_health_index', 'dropout_risk', 'risk_level_encoded', 'gender_Male', 'gender_Other']).columns.tolist()

# --- Common definitions for all panels ---
# Get descriptive statistics to set sensible min/max/default for sliders
descriptive_stats = df_processed.describe().T

# Create a mapping for more descriptive labels
feature_labels = {
    'age': 'Age (Years)',
    'academic_year': 'Academic Year',
    'study_hours_per_day': 'Study Hours Per Day',
    'exam_pressure': 'Exam Pressure Score (0-10)',
    'academic_performance': 'Academic Performance Score (0-100)',
    'stress_level': 'Stress Level Score (0-10)',
    'anxiety_score': 'Anxiety Score (0-10)',
    'depression_score': 'Depression Score (0-10)',
    'sleep_hours': 'Sleep Hours Per Night',
    'physical_activity': 'Physical Activity (Hours/Week)',
    'social_support': 'Social Support Score (0-10)',
    'screen_time': 'Screen Time (Hours/Day)',
    'internet_usage': 'Internet Usage (Hours/Day)',
    'financial_stress': 'Financial Stress Score (0-10)',
    'family_expectation': 'Family Expectation Score (0-10)',
    'burnout_score': 'Burnout Score (0-10)', # Note: This will be an output for panel 3
    'mental_health_index': 'Mental Health Index (0-100)', # Note: This will be an output for panel 2
}

# Helper function to create widgets for a given set of features
def create_widgets_for_panel(feature_list, existing_widgets={}):
    panel_widgets = existing_widgets.copy()
    for feature in feature_list:
        if feature not in panel_widgets and feature in descriptive_stats.index:
            stats = descriptive_stats.loc[feature]
            min_val = stats['min']
            max_val = stats['max']
            mean_val = stats['mean']
            step = (max_val - min_val) / 20.0 if (max_val - min_val) > 0 else 0.1

            panel_widgets[feature] = widgets.FloatSlider(
                value=mean_val,
                min=min_val,
                max=max_val,
                step=step,
                description=feature_labels.get(feature, feature.replace('_', ' ').title()),
                orientation='horizontal',
                continuous_update=False,
                layout=widgets.Layout(width='auto', flex='1 1 auto')
            )
        elif feature not in descriptive_stats.index:
            print(f"Warning: Feature '{feature}' not found in descriptive statistics.")
    return panel_widgets

# --- Panel 1: Predict Risk Level & Dropout Risk ---
input_widgets_panel1 = create_widgets_for_panel(classification_features)
predict_button_panel1 = widgets.Button(description="Predict Risk Level & Dropout Risk", button_style='info')
output_area_panel1 = widgets.Output()
ui_panel1_content = widgets.VBox([input_widgets_panel1[f] for f in classification_features if f in input_widgets_panel1] + [predict_button_panel1, output_area_panel1])

# --- Panel 2: Predict Mental Health Index ---
# Note: mhi_input_features and burnout_input_features are identical, so we can reuse a common set of widgets if desired
# For clarity, creating separately but they'll generate the same sliders.
input_widgets_panel2 = create_widgets_for_panel(mhi_input_features)
predict_button_panel2 = widgets.Button(description="Predict Mental Health Index", button_style='success')
output_area_panel2 = widgets.Output()
ui_panel2_content = widgets.VBox([input_widgets_panel2[f] for f in mhi_input_features if f in input_widgets_panel2] + [predict_button_panel2, output_area_panel2])

# --- Panel 3: Predict Burnout Score ---
input_widgets_panel3 = create_widgets_for_panel(burnout_input_features)
predict_button_panel3 = widgets.Button(description="Predict Burnout Score", button_style='warning')
output_area_panel3 = widgets.Output()
ui_panel3_content = widgets.VBox([input_widgets_panel3[f] for f in burnout_input_features if f in input_widgets_panel3] + [predict_button_panel3, output_area_panel3])
```
